```python
def is_meson_or_antimeson(pdg_code: int) -> bool:
    a = abs(pdg_code)
    if a >= 9900000: return False
    thousands = (a // 1000) % 10
    if thousands != 0: return False
    hundreds = (a // 100) % 10
    if hundreds == 0: return False
    if 901 <= a <= 930: return False
    if a in (110, 990): return False
    if a in (998, 999): return False
    if a == 100: return False
    return True
```
